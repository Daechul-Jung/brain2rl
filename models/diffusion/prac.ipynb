```python
import math
import torch
from typing import Tuple
```

```python
T = torch.arange(1,100 + 1, dtype= torch.float32)
```

```python
T
```

```python
f = torch.cos(((100 / T + 0.0008) / (1 + 0.0008)) * math.pi / 2) **2 
```

```python
alpha_bar = (f/f[0])
```

```python
betas = 1 - (alpha_bar[1:] / alpha_bar[:-1])
betas = betas.clamp(1e-6, 0.999)
```

```python
alpha_bar = torch.cumprod(betas, dim=0)
alpha_bar
```
